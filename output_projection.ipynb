## output projection


```python
import torch
import torch.nn as nn

class OutputLayer(nn.Module):

    def __init__(self,d_model,vocab):

        super().__init__()

        self.linear = nn.Linear(d_model,vocab)

    def forward(self,x):

        return self.linear(x)


# TEST
layer = OutputLayer(16,100)

x = torch.rand(2,5,16)

out = layer(x)

print(out.shape)
```
