## Decoder Stack

```python
import torch
import torch.nn as nn

class Decoder(nn.Module):

    def __init__(self,layers,d_model,heads,d_ff):

        super().__init__()

        self.layers = nn.ModuleList(
            [DecoderLayer(d_model,heads,d_ff)
             for _ in range(layers)]
        )

    def forward(self,x,enc):

        for layer in self.layers:

            x = layer(x,enc)

        return x


# TEST
dec = Decoder(2,16,4,64)

x = torch.rand(2,5,16)
enc = torch.rand(2,5,16)

out = dec(x,enc)

print(out.shape)
```
