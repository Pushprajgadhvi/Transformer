```python
import torch
import torch.nn as nn

# Position-wise Feed Forward Network
# This module is used inside every Transformer encoder/decoder block
class PositionwiseFeedForward(nn.Module):
    def __init__(self, embed_dim, ff_dim):

        # Initialize parent class (required in PyTorch modules)
        super().__init__()

        # First linear layer:
        # Expands feature dimension from embed_dim → ff_dim
        # Example: 4 → 8
        # This increases representational capacity
        self.linear1 = nn.Linear(embed_dim, ff_dim)

        # Non-linear activation function
        # Introduces non-linearity so the model is not purely linear
        self.relu = nn.ReLU()

        # Second linear layer:
        # Projects back from ff_dim → embed_dim
        # Ensures input and output dimensions are the same
        self.linear2 = nn.Linear(ff_dim, embed_dim)

    def forward(self, x):
        """
        x shape: (batch_size, seq_len, embed_dim)

        Important:
        - This operation is applied independently to each token.
        - No interaction between tokens happens here.
        """

        # Step 1: Expand dimension
        # Shape: (batch, seq_len, embed_dim)
        # → (batch, seq_len, ff_dim)
        out1 = self.linear1(x)

        # Step 2: Apply ReLU
        # Negative values become 0
        out2 = self.relu(out1)

        # Step 3: Project back to original dimension
        # Shape: (batch, seq_len, ff_dim)
        # → (batch, seq_len, embed_dim)
        out3 = self.linear2(out2)

        return out3


# -------------------------------
# Reproducibility
# -------------------------------

# Fix random seed so results are same every run
torch.manual_seed(0)

# -------------------------------
# Define dimensions
# -------------------------------

embed_dim = 4   # Size of each token embedding
ff_dim = 8      # Hidden dimension (usually 4× embed_dim in real models)
batch_size = 1
seq_len = 2

# -------------------------------
# Create sample input
# -------------------------------

# Random tensor simulating embeddings
# Shape: (1, 2, 4)
# Meaning:
#   1 batch
#   2 tokens
#   4 features per token
x = torch.randn(batch_size, seq_len, embed_dim)

# -------------------------------
# Initialize FFN
# -------------------------------

ffn = PositionwiseFeedForward(embed_dim, ff_dim)

# -------------------------------
# Forward pass
# -------------------------------

# Pass input through feed-forward network
output = ffn(x)

# -------------------------------
# Print results
# -------------------------------

print("Input:\n", x)
print("\nOutput:\n", output)
print("\nOutput Shape:", output.shape)
```
